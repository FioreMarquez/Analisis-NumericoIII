```python
#Nuevo valor = Valor actual + paso x pendiente

import numpy as np

def euler(f, a, b, y0, dt):
    # Calculamos el número de subintervalos N
    N = int(np.round((b - a) / dt))
    t = np.linspace(a, b, N + 1)
    
    # Arreglo para la solución numérica
    y = np.zeros(N + 1)
    y = y0  # Condición inicial
    
    # Bucle temporal (Forward Euler)
    for j in range(N):
        y[j+1] = y[j] + dt * f(t[j], y[j])
        
    return t, y
```

```python
#inc C) Euler mejorado
def euler_mejorado(f, a, b, y0, dt):
    N = int(np.round((b - a) / dt))
    t = np.linspace(a, b, N + 1)
    y = np.zeros(N + 1)
    y = y0
    
    for j in range(N):
        # Predictor (Euler Explícito)
        y_pred = y[j] + dt * f(t[j], y[j])
        # Corrector (Trapecio)
        y[j+1] = y[j] + (dt / 2.0) * (f(t[j], y[j]) + f(t[j+1], y_pred))
        
    return t, y
```

```python
#
from scipy.optimize import root_scalar

def euler_implicito(f, a, b, y0, dt):
    N = int(np.round((b - a) / dt))
    t = np.linspace(a, b, N + 1)
    y = np.zeros(N + 1)
    y = y0
    
    for j in range(N):
        # Definimos la función g(y_next) = y_next - dt * f(t_{j+1}, y_next) - y[j] = 0
        g = lambda y_next: y_next - dt * f(t[j+1], y_next) - y[j]
        
        # Resolvemos g(y_next) = 0 tomando y[j] como estimación inicial
        sol = root_scalar(g, x0=y[j], x1=y[j] + dt * f(t[j], y[j]))
        y[j+1] = sol.root
        
    return t, y
```

```python
#Inc D)
# Verificación de tasas de convergencia p
dt_list = [0.1, 0.05, 0.025, 0.0125]
f = lambda t, y: y
y_exacta_final = np.exp(2)

def estimar_tasas(metodo_func):
    errores = []
    for dt in dt_list:
        t, y_num = metodo_func(f, 0, 2, 1.0, dt)
        error = np.abs(y_exacta_final - y_num[-1])
        errores.append(error)
        
    p_rates = []
    for i in range(len(errores) - 1):
        p = np.log(errores[i] / errores[i+1]) / np.log(dt_list[i] / dt_list[i+1])
        p_rates.append(p)
    return p_rates

print("Tasa p (Euler Explícito):", estimar_tasas(euler))           # Tiende a 1.0
print("Tasa p (Euler Implícito):", estimar_tasas(euler_implicito)) # Tiende a 1.0
print("Tasa p (Euler Mejorado):", estimar_tasas(euler_mejorado))   # Tiende a 2.0
```
